This final report introduces and analyzes the research on the usage of video game conducted by research group in Computer Science at UBC, led by Frank Wood. The researchers mainly aim to gather data about how and how long people play games, creating MineCraft server that monitors the players' actions. All player activity was automatically logged as users interacted with the virtual world. The data were collected through the game server backend, so it captures objective player behaviour rather than self-reported metrics. During the process, they encountered a problem of how to target recruitment effort and make sure enough resources are prepared instead of indiscriminately collecting players.


With players.csv and sessions.csv, our report gives the best strategies for recruitment effort by answering question 1, which is "What players characteristics and behaviours are most predicative of subscribing to a game-related newsletter, and how do these features differ between various player types?" 
The additional questions are the following:

“Can player demographics such as age, gender, experience level, played_hours and in-game behaviour predict whether a Minecraft research player subscribes to the newsletter?”
